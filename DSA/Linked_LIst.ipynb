```python
#Defining Node in LL
class Node:
    __slots__ = 'element','_next'
    def __init__(self,element,next):
        self.element = element
        self._next = next
n1=Node(7,None)
n2=Node(5,None)
n3=Node(1,None)
```

```python
#Creating a LL
head = None
def CreateLL(x):
    global head
    new_node = Node(x,None)

    if(head==None):
        head = new_node
    else:
        temp = head
        while(temp._next!=None):
            temp=temp._next
        
        temp._next = new_node
    
```

```python
x = int(input('Asking for linked List Input'))

while(x != -1):
    CreateLL(x)
    x = int(input('Asking for linked List Input'))

def show_ll (head):
    temp = head
    while(temp!=None):
        print(f'{temp.element} -> ')
        temp = temp._next
   
show_ll(head)
```
